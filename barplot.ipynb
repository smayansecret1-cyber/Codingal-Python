```python
import seaborn as sns

import matplotlib.pyplot as plt

df = sns.load_dataset('tips')

df = df.dropna()

# Average total bill per day, split by gender

sns.barplot(x='day', y='total_bill', hue='sex', data=df)

plt.title('Average Total Bill per Day by Gender')

plt.xlabel('Day')

plt.ylabel('Average Bill ($)')

plt.show()
```

```python
# Number of diners per day, split by gender

sns.countplot(x='day', hue='sex', data=df)

plt.title('Number of Diners per Day by Gender')

plt.xlabel('Day')

plt.ylabel('Count')

plt.show()
```

```python
# Box plot — spread and outliers per day

sns.boxplot(x='day', y='total_bill', data=df)

plt.title('Spread of Total Bill per Day')

plt.xlabel('Day')

plt.ylabel('Total Bill ($)')

plt.show()
```

```python
# Strip plot — every individual bill shown

sns.stripplot(x='day', y='total_bill', data=df, jitter=True)

plt.title('Every Bill Amount per Day (Strip Plot)')

plt.xlabel('Day')

plt.ylabel('Total Bill ($)')

plt.show()
```

```python
# Swarm plot — every dot spread so none overlap

sns.swarmplot(x='day', y='total_bill', data=df)

plt.title('Every Bill Amount per Day (Swarm Plot)')

plt.xlabel('Day')

plt.ylabel('Total Bill ($)')

plt.show()
```

```python
import seaborn as sns 

import matplotlib.pyplot as plt

df = sns.load_dataset('tips')

df = df.dropna()

print(df.head())

print(df.info())

sns.barplot(x='day', y='total_bill', hue='sex', data=df)

plt.title('Average Total Bill per Day by Gender')

plt.xlabel('Day')

plt.ylabel('Average Bill ($)')

plt.show()

sns.countplot(x='day', hue='sex', data=df)

plt.title('Number of Diners per Day by Gender')

plt.xlabel('Day')

plt.ylabel('Count')

plt.show()
```

```python
sns.jointplot(x='total_bill', y='tip', data=df)
plt.suptitle('Total Bill vs Tip', y=1.02)
plt.show()

sns.jointplot(x='total_bill', y='tip', data=df, kind='kde')
plt.suptitle('Total Bill vs Tip - KDE Joint Plot', y=1.02)
plt.show()
```
