```python
import matplotlib.pyplot as plt
```

```python
days   = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri']
scores = [70, 85, 60, 90, 75]
```

```python
plt.plot(days, scores)
plt.plot(days, scores, color='red')
plt.show()
```

```python
plt.plot(days, scores)
plt.title('My Quiz Score Tracker')
plt.xlabel('Day of the Week')
plt.ylabel('Score')
plt.grid(True)
plt.ylim(0, 100)
plt.plot(days, scores, color='red')
plt.show()
```

```python
plt.plot(days, scores, color='red', marker='o', linestyle='dashed', linewidth=2)
plt.title('My Quiz Score Tracker')
plt.xlabel('Day of the Week')
plt.ylabel('Score')
plt.grid(True)
plt.ylim(0, 100)
plt.show()
```

```python
plt.bar(days, scores, color='red')
plt.title('My Quiz Score Bar Chart')
plt.xlabel('Day of the Week')
plt.ylabel('Score')
plt.ylim(0, 100)
plt.show()
```
